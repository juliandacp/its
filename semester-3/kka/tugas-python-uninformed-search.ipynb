### Soal 1 - DFS

```python
def dfs(graph, start, goal):
    visited = set()
    stack = [start]
    urutan_kunjungan = []
    
    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            urutan_kunjungan.append(node)
            
            if node == goal:
                return urutan_kunjungan

            for tetangga in reversed(graph[node]):
                if tetangga not in visited:
                    stack.append(tetangga)
                    
    return urutan_kunjungan

dfs_result = dfs(graph, 'Arad', 'Bucharest')
print(f"Urutan Kunjungan: {dfs_result}\n")
```

### Soal 2 - BFS

```python
from collections import deque

def bfs(graph, start, goal):
    visited = set()      
    queue = deque([start])
    urutan_kunjungan = []

    visited.add(start)

    while queue:
        node = queue.popleft()
        urutan_kunjungan.append(node)
        
        if node == goal:
            return urutan_kunjungan

        for tetangga in graph[node]:
            if tetangga not in visited:
                visited.add(tetangga)
                queue.append(tetangga)

    return urutan_kunjungan

bfs_result = bfs(graph, 'Arad', 'Bucharest')
print(f"Urutan Kunjungan: {bfs_result}\n")
```

### Soal 3 - UCS

```python
import heapq

def ucs(graph, start, goal):
    visited = set()
    queue = [(0, [start])]
    
    while queue:
        biaya, path = heapq.heappop(queue)
        node = path[-1]
        
        if node in visited:
            continue
            
        visited.add(node)
        
        if node == goal:
            return biaya, path
            
        for tetangga, jarak in graph[node].items():
            if tetangga not in visited:
                heapq.heappush(queue, (biaya + jarak, path + [tetangga]))
                
    return float('inf'), []

ucs_cost, ucs_path = ucs(graph_costs, 'Arad', 'Bucharest')
print(f"Rute Terpendek: {ucs_path}")
print(f"Total Jarak: {ucs_cost} km\n")
```

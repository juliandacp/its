## Kasus Perjalanan dari Ponorogo ke Surabaya
Perjalanan **Ponorogo → Surabaya** menggunakan dua algoritma:
1. **Greedy Best First Search (GBFS)** menggunakan nilai heuristik `h(n)`.
2. **A\*** menggunakan `f(n) = g(n) + h(n)`.

Data jarak antarkota dan jarak garis lurus (heuristik) ditranskripsikan dari gambar soal. Jarak antarkota digunakan sebagai **cost/weight**, sedangkan jarak garis lurus ke Surabaya digunakan sebagai **heuristik**.

```python
# Data Jarak Antarkota
graph = {
    "Ngawi": {"Magetan": 32, "Madiun": 30, "Bojonegoro": 44},
    "Magetan": {"Ngawi": 32, "Madiun": 22, "Ponorogo": 34},
    "Madiun": {"Ngawi": 30, "Magetan": 22, "Ponorogo": 29, "Bojonegoro": 33, "Nganjuk": 48},
    "Ponorogo": {"Magetan": 34, "Madiun": 29},
    "Bojonegoro": {"Ngawi": 44, "Lamongan": 42, "Nganjuk": 70, "Madiun": 33},
    "Nganjuk": {"Madiun": 48, "Bojonegoro": 70, "Jombang": 40},
    "Lamongan": {"Bojonegoro": 42, "Gresik": 14},
    "Gresik": {"Lamongan": 14, "Surabaya": 12, "Bangkalan": 44, "Jombang": 72},
    "Jombang": {"Nganjuk": 40, "Gresik": 72, "Surabaya": 25},
    "Surabaya": {"Gresik": 12, "Jombang": 25, "Sidoarjo": 25},
    "Sidoarjo": {"Surabaya": 25, "Probolinggo": 78},
    "Probolinggo": {"Sidoarjo": 78, "Situbondo": 99},
    "Situbondo": {"Probolinggo": 99},
    "Bangkalan": {"Gresik": 44, "Sampang": 52},
    "Sampang": {"Bangkalan": 52, "Pamekasan": 31},
    "Pamekasan": {"Sampang": 31, "Sumenep": 54},
    "Sumenep": {"Pamekasan": 54}
}

# Heuristik
heuristic = {
    "Magetan": 162,
    "Surabaya": 0,
    "Ngawi": 130,
    "Ponorogo": 128,
    "Madiun": 126,
    "Bojonegoro": 60,
    "Nganjuk": 70,
    "Jombang": 36,
    "Lamongan": 36,
    "Gresik": 12,
    "Sidoarjo": 22,
    "Probolinggo": 70,
    "Situbondo": 146,
    "Bangkalan": 140,
    "Sampang": 90,
    "Pamekasan": 104,
    "Sumenep": 150
}

start = "Ponorogo"
goal = "Surabaya"
```

##### Greedy Best First Search

```python
import heapq

def greedy_best_first_search(graph, heuristic, start, goal):
    # Priority queue: (h(n), urutan, node)
    counter = 0
    frontier = [(heuristic[start], counter, start)]
    parent = {start: None}
    visited = set()
    steps = []

    while frontier:
        h_value, _, current = heapq.heappop(frontier)

        if current in visited:
            continue

        visited.add(current)
        steps.append({
            "Node": current,
            "h(n)": h_value
        })

        if current == goal:
            break

        for neighbor in graph[current]:
            if neighbor not in visited and neighbor not in parent:
                counter += 1
                parent[neighbor] = current
                heapq.heappush(
                    frontier,
                    (heuristic[neighbor], counter, neighbor)
                )

    # Rekonstruksi path
    path = []
    current = goal

    while current is not None:
        path.append(current)
        current = parent.get(current)

    path.reverse()

    # Hitung total jarak
    total_cost = sum(
        graph[a][b] for a, b in zip(path, path[1:])
    )

    return path, total_cost, steps


gbfs_path, gbfs_cost, gbfs_steps = greedy_best_first_search(
    graph, heuristic, start, goal
)

print("Urutan node yang dipilih:")
for i, step in enumerate(gbfs_steps, 1):
    print(f"{i}. {step['Node']:12s} h(n) = {step['h(n)']}")

print("\nPath GBFS:")
print(" -> ".join(gbfs_path))
print(f"Total jarak = {gbfs_cost}")
```
